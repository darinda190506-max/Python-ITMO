```python
def get_name(name):
    if '(' in name:
        name_part = name.split('(')[1]
        return name_part.split()[0].replace('"', '').replace("'", "")
    
    else:
        if '.' in name:
            name_part = name.split('.')[1]
            return name_part.split()[0]
    return name  

df['Clean_Name'] = df['Name'].apply(get_name)

top_male = df[df['Sex'] == 'male']['Clean_Name'].value_counts().index[0]
print(f"Самое популярное мужское имя: {top_male}")

top_female = df[df['Sex'] == 'female']['Clean_Name'].value_counts().index[0]
print(f"Самое популярное женское имя: {top_female}")

top_names_by_class = df.groupby(['Pclass', 'Sex'])['Clean_Name'].agg(lambda x: x.value_counts().index[0])
print(top_names_by_class)
```
