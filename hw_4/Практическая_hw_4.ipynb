**Проинтепретируйте полученный график**

На полученном графике мы можем видеть следующее:
    - Медиана находится медиана находится ближе к 30 годам, что означает что большая часть пассжиров(также основываясь на гистограмме выше) была трудоспособного и молодого возраста 
    - Средний возраст пассажиров составляет около +- 25 лет, что сравнивая с медианой, не дает сильных перекосов в распределении
    - Основная масса пассажиров находилась в возрасте от +-21 до +-38 лет,
    - На графике также присутствуют выбросы в верхней части ящика с усами от 65 до 80 лет, интересно то, что в нижней части ящика с усами подобных выбросов нет, что значит пассажиры от 0 до 20 лет сложились в статистическую норму для данных

**Постройте график типа pie chart для переменных Survived, Pclass, подпишите доли в процентах**

```python
df.groupby(['Survived', 'Pclass']).size().plot(kind='pie', autopct='%1.1f%%')
```

На графике первое значение 0 - выжил, 1 - не выжил, второе значение - номер класса
То есть другими словами, график показывает процент смертности и выживаемости по классам

**Постройте интерактивный sunburst plot (визуализация иерархических данных) с помощью plotly. На первом уровне иерархии - количество пассажиров в каждом из классов, а на втором количество женщин/мужчин в этом классе**
