Для оптимального потребления памяти можно:
- уменьшить разрядность для числовых столбцов:
    - это подойдет для всех столбцов, кроме price, так как в нем хранятся слишком большие значения 
    - поэтому я заменю потребление памяти с int(64) до int(16)
- совершить переход к категориальным значениям для строковых значений
    - в данном случае в каждом строковом столбце содержатся данные которые можно переписать как категориальные 

Поменяйте типы данных столбцов датафрейма на выбранные вами в прошлом пункте и сравните потребление памяти до и после оптимизации
